### Classes and objects crash course

Dog is a class. Think of it as a template for creating an object.

```python
class Dog:
    def __init__(self, name, age):
        """ 
        This is the intializing function.
        It runs and sets up internal variables when
        the object is created
        """
        self.name = name
        self.age = age

    def bark(self):
        print(self.name, "says woof!")

    def call(self, what_you_said):
        if self.name in what_you_said:
            print(self.name, "says woof!")
        else:
            print('silence.')

    def birthday(self):
        self.age += 1
```

dog1 is an object we created with the class. It is an *instance* of Dog.

```python
dog1 = Dog('Milo', 1)
```

Run the following lines of code and see what happens. If you see an error, think about why.
