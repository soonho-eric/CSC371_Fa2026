### 8 Puzzle Problem
Let's check if our functions work for a different problem, the 8 puzzle.

````python
class EightPuzzleProblem:
    """
    Eight-puzzle search problem.

    States are strings of length 9 containing the digits 0 through 8.
    The digit 0 represents the blank space.

    Example:
        "123405786"

    represents

        1 2 3
        4   5
        7 8 6
    """

    GOAL = "123456780"

    def __init__(self, initial):
        """
        Parameters
        ----------
        initial : str
            A 9-character string containing each digit 0 through 8 once.
        """

        if len(initial) != 9:
            raise ValueError("Initial state must contain exactly 9 characters.")

        if set(initial) != set("012345678"):
            raise ValueError(
                "Initial state must contain each digit 0 through 8 exactly once."
            )

        self.start = initial


    def goal_test(self, state):
        """
        Return True if state is the goal state.
        """

    def neighbors(self, state):
        """
        Return all states that can be reached by moving the blank
        one square up, down, left, or right.
        """

        blank = state.index("0")

        row = blank // 3
        col = blank % 3

        possible_moves = []

        if row > 0:# up
            possible_moves.append(blank - 3)

        if row < 2: # down
            possible_moves.append(blank + 3)

        if col > 0: # left
            possible_moves.append(blank - 1)

        if col < 2: # right
            possible_moves.append(blank + 1)

        neighbors = []

        for new_blank in possible_moves:

            new_state = list(state)

            new_state[blank], new_state[new_blank] = (
                new_state[new_blank],
                new_state[blank]
            )

            neighbors.append("".join(new_state))

        return neighbors


    def display(self, state):
        for row in range(3):

            start = 3 * row
            end = start + 3

            cells = state[start:end]

            cells = [" " if value == "0" else value for value in cells]
            print("| " + " | ".join(cells) + " |")


    def display_path(self, path):
        """
        Print every state in a solution path.
        """

        for step, state in enumerate(path):

            print(f"Step {step}")
            self.display(state)
            print()
````

````python
problem = EightPuzzleProblem("123405786")

goal, path, expanded = bfs(problem)
````
