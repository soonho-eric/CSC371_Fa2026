Let's start by defining two vectors.

````python
u = np.array([1, 2])
v = np.array([-3, 3])
````

Try evaluating each of the following and identify what they do.

````python
# visualize u
fig, ax = plt.subplots()
ax.quiver(0, 0, u[0], u[1],
          angles="xy", scale_units="xy", scale=1,
          color="blue")

ax.axhline(0, color="gray", linewidth=0.5)
ax.axvline(0, color="gray", linewidth=0.5)
ax.set_xlim(-5,5)
ax.set_ylim(-5,5)
ax.set_aspect("equal")

plt.show()
````

````python
import numpy as np
import matplotlib.pyplot as plt

theta = np.array([2, 1])
b = 1

x, y = np.meshgrid(np.linspace(-4, 4, 100),
                   np.linspace(-4, 4, 100))

plt.contour(x, y, theta[0]*x + theta[1]*y - b,
            levels=[0], colors="black")
plt.quiver(0, 0, *theta, angles="xy", scale_units="xy", scale=1)

plt.axis("equal")
plt.xlim(-4, 4)
plt.ylim(-4, 4)
plt.grid()
plt.show()
````
