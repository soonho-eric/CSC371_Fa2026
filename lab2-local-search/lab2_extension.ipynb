## A. The thermodynamics of annealing

Simulated annealing borrows its name and its rule from statistical physics. The borrowing goes further than the name: run the Metropolis rule at a *fixed* temperature $T$ long enough and the states you visit are distributed as $P(s) \propto e^{-L(s)/T}$, where $L$ is the tour length. So a fixed-temperature run is a sampler, and you can measure things.

The quantity to measure is the **heat capacity**

$$C(T) = \frac{\operatorname{Var}[L]}{T^2}.$$
